## Overview

This project will investigate how changes in the federal funds rate influence U.S. stock market returns and volatility. Using historical federal interest rates and S&P 500 index data, the project quantifies whether rate hikes, cuts, or the absolute level of rates can explain subsequent market performance and risk, similar to how Wall Street analysts monitor macroeconomic signals to guide investment decisions.

## Research Questions

Do changes in the federal funds rate predict short-term or medium-term stock market returns (1-month, 3-month, 6-month)?

How do rate changes impact realized market volatility over similar horizons?

Are the effects of rate changes dependent on rate regimes (rising vs falling, high vs low rates)?

## Integration Plan

Aggregate both datasets to monthly frequency for alignment.

Compute metrics:
Rate change over the month.
Rate level (absolute rate).
Market return (monthly return of S&P 500).
Market volatility (standard deviation of daily returns in the month).

Combine into a single panel dataset: month, rate_level, rate_change, market_return, market_volatility.

## Constraints

There are several constraints that will impact the execution as well as interpretation of this project. First, the frequency mismatch between federal funds rate data and stock market data must be carefully managed because of the fact that the federal funds rate is reported daily or monthly, while stock returns can be calculated daily, weekly, or monthly. This creates a problem that will needs to be fixed when I build this code. Aligning these two datasets requires aggregation and/or ‘smoothing’ of the data, which may reduce precision or obscure these short-term relationships. Secondly, the causal relationship between interest rates and stock market performance is a bit complex. Rate changes themselves are influenced by economic conditions, such as inflation and employment trends, which also additionally directly affect stock prices. As a result, almost any observed correlation between rate changes and market returns could reflect underlying macroeconomic conditions rather than a direct causal effect which is what I am going for. Third, while the datasets chosen provide quite a bit of historical coverage, there is a limit to the information they contain. Additional macroeconomic variables such as GDP growth, consumer confidence, or corporate earnings could provide important context but are not included in this initial analysis (however may be included if I find it valuable through the course of the project). Finally, external shocks, such as financial crises, geopolitical events, or pandemic-related market disruptions, can have outsized effects on returns and volatility, making it difficult to generalize results across different time periods.


## Known Gaps / Needs

Several gaps/areas requiring more input have been identified. These will influence the depth and the rigor of the project. One primary concern is the alignment of federal rate announcement dates with the calculation of stock returns. For example, a rate change that gets announced mid-month may have an immediate impact on market sentiment, but a monthly aggregation could dilute this effect. Determining the proper lag structure for rate changes and making sure they align with market data is very essential for meaningful analysis, especially for this specific project. Another gap involves the limited set of explanatory variables included in the analysis. While rate changes and rate levels are the primary independent variables, other macroeconomic indicators such as inflation, unemployment, or industrial production significantly influence market returns and volatility. Including these variables would strengthen the results of the project. Additionally, the impact of different interest rate regimes - such as periods of rising rates, falling rates, or historically low rates - need to be modeled explicitly using dummy variables or some other technique. Finally, there is likely to be data quality issues, such as missing values or revisions to historical rate series. This will most definitely require careful and detailed handling to avoid introducing bias or errors in the final analysis of this.

This project anticipates the incorporation of many advanced topics that will have to be addressed in our coursework for later weeks. Regression analysis will be a forefront part of this project, specifically in testing the predictive power of rate changes and rate levels on future stock returns and volatility. Additionally, this will require careful consideration of multicollinearity, lagged variables, and the potential problem that the variance of the residuals (or the error terms) will not be constant across all levels of an independent variable in the residuals. I also believe that time series modeling is another anticipated area, this includes techniques such as ARIMA or Vector Autoregression (VAR) to highlight relationships between interest rates and market returns over a given period of time. Feature engineering will be important for capturing relevant signals, like for example rolling averages of rate changes, volatility measures, and regime indicators for rising or falling interest rate environments. Additionally, advanced visualization techniques will be created to show the causal relationships between rates and market performance. This will include scatter plots, lagged correlation heatmaps, and time-series overlays if everything goes according to plan. By planning for these topics in advance, the project is set up to be structured as flexible as well as scalable, allowing for enhancements in the future when needed and conforming with more sophisticated modeling techniques introduced later in the course as I see fit. Finally, the project anticipates incorporating checks and sensitivity analyses to validate the data-driven findings, including alternative methods, outlier removal, and comparisons across different time horizons (as with regular stocks), this will do good in ensuring the results are interpretable and replicable by other data scientists.
